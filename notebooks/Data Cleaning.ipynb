```python
import pandas as pd
```

# Human Data

```python
df1 = pd.read_csv("../data/iowa_human/combined_igt_human.csv")
```

```python
df1
```

```python
df1.shape
```

```python
df1.groupby("variant")["session_id"].count()
```

### Checking for missing sessions and trials

```python
human_grouped_by_variant_session = df1.groupby(['variant', 'session_id']).agg(
    num_trials=('trial_number', 'count')
).reset_index()

human_grouped_by_variant_session.head()
```

```python
# For Variant 1:
human_variant1_sessions = human_grouped_by_variant_session[human_grouped_by_variant_session['variant'] == 1]
human_unique_sessions_v1 = human_variant1_sessions['session_id'].nunique()
human_trials_per_session_v1 = human_variant1_sessions['num_trials'].value_counts().reset_index()
human_trials_per_session_v1.columns = ['num_trials', 'count']

print(f"Number of unique sessions in Variant 1: {human_unique_sessions_v1}")
print(f"Distribution of trials per session in Variant 1:\n{human_trials_per_session_v1}")
```

```python
# For Variant 2
human_grouped_by_exp_variant_session = df1.groupby(['experiment', 'variant', 'session_id']).size().reset_index(name='num_trials')

human_variant2_sessions = human_grouped_by_exp_variant_session[human_grouped_by_exp_variant_session['variant'] == 2]
human_unique_sessions_v2 = human_variant2_sessions.shape[0]
human_trials_per_session_v2 = human_variant2_sessions['num_trials'].value_counts().reset_index()
human_trials_per_session_v2.columns = ['num_trials', 'count']

print(f"Number of unique sessions in Variant 2: {human_unique_sessions_v2}")
print(f"Distribution of trials per session in Variant 2:\n{human_trials_per_session_v2}")
```

```python
# For Variant 3:
human_variant3_sessions = human_grouped_by_variant_session[human_grouped_by_variant_session['variant'] == 3]
human_unique_sessions_v3 = human_variant3_sessions['session_id'].nunique()
human_trials_per_session_v3 = human_variant3_sessions['num_trials'].value_counts().reset_index()
human_trials_per_session_v3.columns = ['num_trials', 'count']

print(f"Number of unique sessions in Variant 3: {human_unique_sessions_v3}")
print(f"Distribution of trials per session in Variant 3:\n{human_trials_per_session_v3}")
```

### Checking for missing values and duplicates

```python
df1.info()
```

```python
df1.duplicated().sum()
```
